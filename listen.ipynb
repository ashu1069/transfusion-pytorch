```python
# Load the LibriTTS dataset from local cache
from datasets import load_dataset
import IPython.display as ipd

# Load the dev.clean split (smaller for quick exploration)
dataset = load_dataset(
    "mythicinfinity/libritts",
    "clean",
    split="dev.clean",
    cache_dir="./data_libritts"
)

print(f"Dataset size: {len(dataset)} samples")
print(f"Features: {dataset.features}")
print(f"Sample rate: {dataset.features['audio'].sampling_rate} Hz")
```

```python
# Listen to a few random samples
import random

random.seed(42)
sample_indices = random.sample(range(len(dataset)), 5)

for i, idx in enumerate(sample_indices):
    sample = dataset[idx]
    print(f"\n--- Sample {i+1} (index {idx}) ---")
    print(f"Speaker ID: {sample['speaker_id']}")
    print(f"Text: {sample['text_normalized']}")
    print(f"Duration: {len(sample['audio']['array']) / sample['audio']['sampling_rate']:.2f}s")
    display(ipd.Audio(sample['audio']['array'], rate=sample['audio']['sampling_rate']))
```

```python
# Listen to a specific sample by index
def play_sample(idx):
    """Play a sample from the dataset by index"""
    sample = dataset[idx]
    print(f"Speaker ID: {sample['speaker_id']}")
    print(f"Original text: {sample['text_original']}")
    print(f"Normalized text: {sample['text_normalized']}")
    print(f"Duration: {len(sample['audio']['array']) / sample['audio']['sampling_rate']:.2f}s")
    return ipd.Audio(sample['audio']['array'], rate=sample['audio']['sampling_rate'])

# Try it out - change the index to explore different samples
play_sample(0)
```
